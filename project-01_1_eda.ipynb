```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

numerical_features = [
    "Avg. Area Income",
    "Avg. Area House Age",
    "Avg. Area Number of Rooms",
    "Avg. Area Number of Bedrooms",
    "Area Population",
    "Price"
]

correlation = df[numerical_features].corr()

print(correlation)
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Features")
plt.show()
```

```python
price_correlation = correlation["Price"].sort_values(ascending=False)

print(price_correlation)
```

```python
price_corr = correlation["Price"].drop("Price")

strongest_feature = price_corr.abs().idxmax()
strongest_value = price_corr[strongest_feature]

print("Strongest feature:", strongest_feature)
print("Correlation with Price:", strongest_value)
```

```python
import matplotlib.pyplot as plt

features = [
    "Avg. Area Income",
    "Avg. Area House Age",
    "Avg. Area Number of Rooms",
    "Avg. Area Number of Bedrooms",
    "Area Population"
]

for feature in features:
    plt.figure(figsize=(8, 5))
    
    plt.scatter(df[feature], df["Price"], alpha=0.5)
    
    plt.xlabel(feature)
    plt.ylabel("Price")
    plt.title(f"Price vs {feature}")
    
    plt.grid(True, alpha=0.3)
    plt.show()
```

Key Findings: The exploratory analysis indicates that Avg. Area Income is one of the most important variables associated with house Price, while other features such as house age, area population, and number of rooms also provide useful predictive information. Most relationships appear approximately linear, although some features may contain non-linear patterns. Therefore, all numerical features will initially be considered for modeling.
